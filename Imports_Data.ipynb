We can Handle missing values with imputation using sklrearn

```python
from numpy._core.fromnumeric import mean
from sklearn.impute import SimpleImputer
```

Imputation for quantative

```python
result = SimpleImputer(strategy = 'mean')
df[quantative] = result.fit_transform(df[quantative])
```

Imputation for Categorical 

```python
result2 = SimpleImputer(strategy = 'most_frequent')
df[categorical] = result2.fit_transform(df[categorical])
```

Remove Duplicates

```python
df.drop_duplicates(inplace=True)
```

Handle The Outliers 

```python
Q1 = df['engine-size'].quantile(0.25)
Q3 = df['engine-size'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 +  1.5 * IQR
df= df[(df ['engine-size'] >= lower_bound)&(df['engine-size']<= upper_bound)]
```
